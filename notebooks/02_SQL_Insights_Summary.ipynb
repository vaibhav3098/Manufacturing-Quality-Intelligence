## Key Findings from SQL

### Quality Analysis

- Stage 2 Measurement 4 showed the **highest average deviation** from its target setpoint.
- This measurement was selected as the **Critical-to-Quality (CTQ)** characteristic for further investigation, final parameter for model training to be updated if selected target variable found nt suitable for model training.

### Process Analysis

The highest process variability was observed in:

- Machine 1 Raw Material Property 3
- Machine 1 Motor Current
- Machine 1 Raw Material Feeder
- Machine 4 Pressure
- Machine 1 Material Pressure

### Process Driver Analysis

**Strong Association**

- Machine 1 Raw Material Property 3
- Machine 4 Pressure

**Weak Association**

- Machine 1 Motor Current
- Machine 1 Raw Material Feeder
- Machine 1 Material Pressure
